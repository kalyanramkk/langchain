**Setting Up LLM using API Key**

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from google.colab import userdata
os.environ["GOOGLE_API_KEY"] =  userdata.get("GEMINI_API_KEY")
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash",
      temperature=0,timeout=30,max_retries=2)

```

**Import Required modules from Langchain**

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

**Create Prompt Specification and Persona**

```python
system_prompt_instructions = """

You are a Prompt Quality Scoring Agent who takes the user prompt as an input and evaluate the prompt quality between 0 and 10.


Quality Criterion For Scoring

1. Clarity of prompt with score between 0 and 10 which checks whether the prompt is easy to understand and has a clear goal
2. Specificity/ Details with score between 0 and 10 which evaluates whether sufficient details and requirements are provided.
3. Context with score between 0 and 10 which Checks if background information, audience, or use case is mentioned.
4. Output Format and constraints between 0 and 10 which Checks whether expected output format, tone, or length is specified.
5. Confirm if a prompt assigns a specific role and assign value between 0 and 10

Output for the prompt should contain

1. Final Score between 0 and 10 which must be the average of the above criteria
2. Scores for each Quality Criterion with short explanation
3. 2-3 suggestions for improving the prompt

"""

prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system",system_prompt_instructions),
        ("human","{EvaluationPrompt}")

    ])
```

**Create a variable to store user prompt**


```python
travel_agent_prompt = """ You are Travel agent who plans vacations to destination in India. you take the input for destination
and number of vacation and plan the itenery"""
```

```python
prompt_value = prompt_template.format_prompt(EvaluationPrompt = travel_agent_prompt)
```

**Response**

```python
response = llm.invoke(prompt_value)
```

```python
parser = StrOutputParser()
result = parser.invoke(response)
print(result)
```
